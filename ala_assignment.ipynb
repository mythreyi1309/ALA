```python
from typing import Self
import random
import math
#VECTOR CLASS Creation

class vec:
    def __init__(self,src=None)->Self:
      if src is None :
        self.elements=[]
      else:
        elements=list(src)
        for x in elements:
          if not isinstance(x,(int,float)):
            raise TypeError("Values must be scalar numbers ")
        self.elements=elements


    def __add__(self,t:Self)->Self:
      if not isinstance(t,vec):
        raise TypeError("The list type should be vector  ")
      if len(self.elements)!=len(t):
        raise TypeError("Dimensions should be same ")
      return vec(x+y for x,y in zip(self.elements,t.elements))

    def __rmul__(self,scalar:int | float)->Self:
      if not isinstance(scalar,(int,float)):
        raise TypeError("Invalid value for scalar")
      return vec(scalar*x for x in self.elements)

    def __imul__(self,scalar:int|float) ->Self:
      if not isinstance(scalar,(int,float)):
        raise TypeError("scalar value should be either type int or float")
      for i,val in enumerate(self.elements):
        self.elements[i]=round(val*scalar,5)
      return self

    def __repr__(self):
      return repr(self.elements)

    def __len__(self):
      return len(self.elements)

    def __sub__(self,t:Self)->Self:
      if not isinstance(t,vec):
        raise TypeError("The list must be of type vector ")
      if len(self.elements)!=len(t.elements):
        raise TypeError("The length of both vector must be same ")
      return vec([x-y for x,y in zip(self.elements,t.elements)])

    def __neg__(self):
      for i,val in enumerate(self.elements):
        self.elements[i]=-1*val
      return self

    def __radd__(self, other):
        if not isinstance(other, vec):
            raise TypeError("The list must be of type vector")

        return vec((x + y for x, y in zip(self.elements, other.elements)))


    def __iadd__(self, other):
       if not isinstance(other, vec):
          raise TypeError("The list must be of type vector")
       if len(self) != len(other):
          raise TypeError("The length of both vectors r the same ")
       self.elements = [x+y for x,y in zip(self.elements, other.elements)]
       return self

    @staticmethod
    def zeros(n: int) -> Self:
      v=[0]*n
      return vec(v)

    @staticmethod
    def ones(n: int) -> Self:
      v=[1]*n
      return vec(v)

    @staticmethod
    def uniform(n: int) -> Self:
        v=vec(random.random() for i in range(n))
        return v


    def norm(self)->float:
      sum=0
      for i in self.elements:
        square=i*i
        sum=sum+square

      return math.sqrt(sum)


    def mean(self) -> float:
        if len(self.elements) == 0:
            raise ValueError("Cannot calculate mean of an empty vector")
        return sum(self.elements) / len(self.elements)

    def demean(self) -> Self:
        mean_value = self.mean()
        return vec(x - mean_value for x in self.elements)

    def std(self) -> float:
        demeaned = self.demean()
        return math.sqrt(
            sum(x * x for x in demeaned.elements) / len(self.elements)
        )
```

```python
#Test add
v1=vec([1,2,3,4])
v2=vec([10,20,30,40])
result=vec.__add__(v1,v2)
print(result)
assert result.elements==[11,22,33,44]


#Test rmul
v3=vec([1,2,3,4,5,6])
r=3
rmul=v3.__rmul__(r)
print(rmul)
assert rmul.elements==[3,6,9,12,15,18]

#Test imul
v4=vec([1,2,3,4,5,6,7])
i=2
v4.__imul__(2)
print(v4)
assert v4.elements==[2,4,6,8,10,12,14]

#Test len
print(vec.__len__(v4))

#Test sub
difference=vec.__sub__(v2,v1)
print(difference)
assert difference.elements==[9,18,27,36]


#Test neg
assert vec.__neg__(difference).elements==[-9,-18,-27,-36]

#Test radd
re1=vec.__radd__(v1,v2)
print(re1)
assert re1.elements==[11,22,33,44]

#Test iadd
vec.__iadd__(v1,v2)
print(v1)
assert v1.elements==[11,22,33,44]


#Test zeros
print(vec.zeros(3))

#Test ones
print(vec.ones(3))

#Test uniform
print(vec.uniform(5))

#Test norm [Euclidean's space]
norm=vec.norm(vec([1,2,3,4]))
print(norm)
assert norm==5.477225575051661

#Test mean
v5=vec([1,2,3,4,5])
mean=vec.mean(v5)
print(mean)
assert mean==3.0


v6=vec([-2,0,2])
mean=vec.mean(v6)
print(mean)
assert mean==0.0



v7=vec([10])
mean=vec.mean(v7)
print(mean)
assert mean==10.0



v8=vec([1,2,3,4,5])
demeaned=vec.demean(v8)
print(demeaned)
assert demeaned.elements==[-2.0,-1.0,0.0,1.0,2.0]



v9=vec([5,5,5,5])
demeaned=vec.demean(v9)
print(demeaned)
assert demeaned.elements==[0.0,0.0,0.0,0.0]



v10=vec([-2,0,2])
demeaned=vec.demean(v10)
print(demeaned)
assert demeaned.elements==[-2.0,0.0,2.0]


v11=vec([1,2,3,4,5])
std=vec.std(v11)
print(std)
assert std==math.sqrt(2)



v12=vec([5,5,5,5])
std=vec.std(v12)
print(std)
assert std==0.0


#Test std with negative values
v13=vec([-2,0,2])
std=vec.std(v13)
print(std)
assert std==math.sqrt(8/3)



v14=vec([])

try:
    mean=vec.mean(v14)
    assert False
except ValueError:
    print("Mean empty vector: ValueError passed")

```
