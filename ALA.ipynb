```python
from typing import Self
import random
import math
#VECTOR CLASS Creation

class vec:
    def __init__(self,src=None)->Self:
      if src is None :
        self.elements=[]
      else:
        elements=list(src)
        for x in elements:
          if not isinstance(x,(int,float)):
            raise TypeError("Values must be scalar numbers ")
        self.elements=elements


    def __add__(self,t:Self)->Self:
      if not isinstance(t,vec):
        raise TypeError("The list type should be vector  ")
      if len(self.elements)!=len(t):
        raise TypeError("Dimensions should be same ")
      return vec(x+y for x,y in zip(self.elements,t.elements))

    def __rmul__(self,scalar:int | float)->Self:
      if not isinstance(scalar,(int,float)):
        raise TypeError("Invalid value for scalar")
      return vec(scalar*x for x in self.elements)

    def __imul__(self,scalar:int|float) ->Self:
      if not isinstance(scalar,(int,float)):
        raise TypeError("scalar value should be either type int or float")
      for i,val in enumerate(self.elements):
        self.elements[i]=round(val*scalar,5)
      return self

    def __repr__(self):
      return repr(self.elements)

    def __len__(self):
      return len(self.elements)

    def __sub__(self,t:Self)->Self:
      if not isinstance(t,vec):
        raise TypeError("The list must be of type vector ")
      if len(self.elements)!=len(t.elements):
        raise TypeError("The length of both vector must be same ")
      return vec([x-y for x,y in zip(self.elements,t.elements)])

    def __neg__(self):
      for i,val in enumerate(self.elements):
        self.elements[i]=-1*val
      return self

    def __radd__(self, other):
        if not isinstance(other, vec):
            raise TypeError("The list must be of type vector")

        return vec((x + y for x, y in zip(self.elements, other.elements)))


    def __iadd__(self, other):
       if not isinstance(other, vec):
          raise TypeError("The list must be of type vector")
       if len(self) != len(other):
          raise TypeError("The length of both vectors r the same ")
       self.elements = [x+y for x,y in zip(self.elements, other.elements)]
       return self

    @staticmethod
    def zeros(n: int) -> Self:
      v=[0]*n
      return vec(v)

    @staticmethod
    def ones(n: int) -> Self:
      v=[1]*n
      return vec(v)

    @staticmethod
    def uniform(n: int) -> Self:
        v=vec(random.random() for i in range(n))
        return v


    def norm(self)->float:
      sum=0
      for i in self.elements:
        square=i*i
        sum=sum+square

      return math.sqrt(sum)


    def mean(self) -> float:
        if len(self.elements) == 0:
            raise ValueError("Cannot calculate mean of an empty vector")
        return sum(self.elements) / len(self.elements)

    def demean(self) -> Self:
        mean_value = self.mean()
        return vec(x - mean_value for x in self.elements)

    def std(self) -> float:
        demeaned = self.demean()
        return math.sqrt(
            sum(x * x for x in demeaned.elements) / len(self.elements)
        )
```

```python
import time

sizes = [2000, 4000, 8000, 16000, 32000, 64000]

for n in sizes:

    print("Size:", n)

    v1 = vec.uniform(n)
    v2 = vec.uniform(n)

    # Addition
    start = time.time()
    result = v1 + v2
    end = time.time()
    print("Addition:", end - start)

    # Subtraction
    start = time.time()
    result = v1 - v2
    end = time.time()
    print("Subtraction:", end - start)

    # Scalar multiplication
    start = time.time()
    result = 2 * v1
    end = time.time()
    print("Scalar multiplication:", end - start)

    # In-place multiplication
    v3 = vec(v1.elements)

    start = time.time()
    v3 *= 2
    end = time.time()
    print("In-place multiplication:", end - start)

    # Norm
    start = time.time()
    result = v1.norm()
    end = time.time()
    print("Norm:", end - start)

    # Zeros
    start = time.time()
    result = vec.zeros(n)
    end = time.time()
    print("Zeros:", end - start)

    # Ones
    start = time.time()
    result = vec.ones(n)
    end = time.time()
    print("Ones:", end - start)

    # Uniform random vector
    start = time.time()
    result = vec.uniform(n)
    end = time.time()
    print("Uniform:", end - start)

    print("------------------------")



```

```python
import numpy as np
import time

sizes = [2000, 4000, 8000, 16000, 32000, 64000]

for n in sizes:

    print("Size:", n)

    v1 = np.random.random(n)
    v2 = np.random.random(n)

    # Addition
    start = time.time()
    result = v1 + v2
    end = time.time()
    print("Addition:", end - start)

    # Subtraction
    start = time.time()
    result = v1 - v2
    end = time.time()
    print("Subtraction:", end - start)

    # Scalar multiplication
    start = time.time()
    result = 2 * v1
    end = time.time()
    print("Scalar multiplication:", end - start)

    # In-place multiplication
    v3 = v1.copy()

    start = time.time()
    v3 *= 2
    end = time.time()
    print("In-place multiplication:", end - start)

    # Norm
    start = time.time()
    result = np.linalg.norm(v1)
    end = time.time()
    print("Norm:", end - start)

    # Zeros
    start = time.time()
    result = np.zeros(n)
    end = time.time()
    print("Zeros:", end - start)

    # Ones
    start = time.time()
    result = np.ones(n)
    end = time.time()
    print("Ones:", end - start)

    # Uniform random vector
    start = time.time()
    result = np.random.random(n)
    end = time.time()
    print("Uniform:", end - start)

    print("------------------------")
```
